```python
리스트 = [1,2,3,4,5]
print(리스트[0])
리스트[2] = 200
print(리스트)
리스트.append(6)
print(리스트)
리스트.insert(1, 100)
print(리스트)
리스트.remove(4)
print(리스트)
리스트.pop(2)
print(리스트)
del 리스트[0]
print(리스트)
print(len(리스트))
리스트.sort()
print(리스트)
리스트.sort(reverse=True)
print(리스트)
리스트.count(5)
개수=리스트.count(5)
print(개수)
```
